## Python Operators

### 1. Arithmetic Operators

```python
a = 10
b = 3

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Floor Division:", a // b)
print("Modulus:", a % b)
print("Exponent:", a ** b)
```

---

### 2. Comparison Operators

```python
a = 10
b = 5

print("Equal:", a == b)
print("Not Equal:", a != b)
print("Greater Than:", a > b)
print("Less Than:", a < b)
print("Greater or Equal:", a >= b)
print("Less or Equal:", a <= b)
```

---

### 3. Assignment Operators

```python
x = 5
print("Initial:", x)

x += 3
print("Add and assign:", x)

x -= 2
print("Subtract and assign:", x)

x *= 4
print("Multiply and assign:", x)

x /= 2
print("Divide and assign:", x)

x //= 2
print("Floor divide and assign:", x)

x %= 3
print("Modulus and assign:", x)

x **= 2
print("Exponent and assign:", x)
```

---

### 4. Logical Operators

```python
a = True
b = False

print("AND:", a and b)
print("OR:", a or b)
print("NOT:", not a)
```

---

### 5. Bitwise Operators

```python
a = 5
b = 3

print("AND:", a & b)
print("OR:", a | b)
print("XOR:", a ^ b)
print("NOT:", ~a)
print("Left Shift:", a << 1)
print("Right Shift:", a >> 1)
```

---

### 6. Membership Operators

```python
text = "python"

print("p in text:", "p" in text)
print("z not in text:", "z" not in text)
```

---

### 7. Identity Operators

```python
a = [1, 2, 3]
b = a
c = [1, 2, 3]

print("a is b:", a is b)
print("a is c:", a is c)
print("a is not c:", a is not c)
```

---

### 8. Conditional (Ternary) Operator

```python
age = 20

result = "Adult" if age >= 18 else "Minor"
print(result)
```
