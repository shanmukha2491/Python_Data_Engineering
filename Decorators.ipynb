# Python Decorators

A **decorator** in Python is a function that **modifies or extends the behavior of another function without changing its source code**.

Decorators are commonly used for:

- Logging
- Authentication
- Timing functions
- Caching
- Access control

---

# Basic Concept

A decorator takes a **function as input** and returns a **new function with added behavior**.

```
function → decorator → enhanced function
```

---

# Example Without Decorator

```python
def greet():
    print("Hello!")

def wrapper(func):
    print("Before function call")
    func()
    print("After function call")

wrapper(greet)
```

Output

```
Before function call
Hello!
After function call
```

---

# Creating a Decorator

```python
def my_decorator(func):

    def wrapper():
        print("Before function runs")
        func()
        print("After function runs")

    return wrapper
```

---

# Using Decorator (Manual Way)

```python
def say_hello():
    print("Hello")

say_hello = my_decorator(say_hello)

say_hello()
```

Output

```
Before function runs
Hello
After function runs
```

---

# Using `@` Syntax (Decorator Syntax)

Python provides a cleaner syntax for decorators.

```python
def my_decorator(func):

    def wrapper():
        print("Before function runs")
        func()
        print("After function runs")

    return wrapper


@my_decorator
def say_hello():
    print("Hello")

say_hello()
```

Equivalent to:

```
say_hello = my_decorator(say_hello)
```

---

# Decorator With Arguments

To allow decorated functions to accept parameters, use `*args` and `**kwargs`.

```python
def my_decorator(func):

    def wrapper(*args, **kwargs):
        print("Function starting")
        result = func(*args, **kwargs)
        print("Function finished")
        return result

    return wrapper


@my_decorator
def add(a, b):
    return a + b


print(add(3, 5))
```

Output

```
Function starting
Function finished
8
```

# Real Example: Execution Time Decorator

```python
import time

def timer(func):

    def wrapper(*args, **kwargs):
        start = time.time()
        result = func(*args, **kwargs)
        end = time.time()

        print("Execution time:", end - start)
        return result

    return wrapper


@timer
def slow_function():
    time.sleep(2)
    print("Task completed")


slow_function()
```

# Common Built-in Decorators

Python provides several built-in decorators.

### `@staticmethod`

```python
class Math:

    @staticmethod
    def add(a, b):
        return a + b
```

### `@classmethod`

```python
class Example:

    count = 0

    @classmethod
    def increment(cls):
        cls.count += 1
```

### `@property`

```python
class Person:

    def __init__(self, name):
        self._name = name

    @property
    def name(self):
        return self._name
```
